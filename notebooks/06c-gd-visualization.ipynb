## Part 1: Gradient Descent for Linear Regression

We fit a simple linear model $y = \omega x + b + \varepsilon$ using gradient descent. The loss surface for this 2-parameter model is a convex paraboloid, so GD converges to the global minimum. Below we implement a basic GD loop that computes the partial derivatives $\partial L/\partial \omega$ and $\partial L/\partial b$ analytically and takes steps of size $\eta$.

The model is $y = \omega x + b + \varepsilon$, and we minimize the mean squared error:

$$
L(\omega, b) = \frac{1}{n}\sum_{i=1}^n (y_i - \omega x_i - b)^2.
$$

### One-dimensional gradient descent

The gradient update rules are:

$$
\omega \leftarrow \omega - \eta \cdot \frac{1}{n}\sum_i -x_i(y_i - \omega x_i - b), \qquad
b \leftarrow b - \eta \cdot \frac{1}{n}\sum_i -(y_i - \omega x_i - b).
$$

```python
def gradient_descent(x, y, eta=1e-3, epochs=1e5):
    omega_t, b_t = 0,0
    for epoch in range(int(epochs)):
        # calculate gradient
        dw = np.mean(-x*(y-omega_t*x-b_t))
        db = np.mean(-(y-omega_t*x-b_t))

        # update coefficients
        omega_t = omega_t - dw*eta
        b_t = b_t - db*eta
    return({"omega": omega_t, "b": b_t})
```

## Part 2: Newton-Raphson Method

Newton's method uses second-order (Hessian) information for faster convergence. The update is

$$
\theta^{(k+1)} = \theta^{(k)} - \bigl[\nabla^2 L(\theta^{(k)})\bigr]^{-1} \nabla L(\theta^{(k)}).
$$

For the MSE loss with a linear model the Hessian is constant ($H = \frac{2}{n}X^\top X$), so Newton's method converges in one step. This is in contrast to gradient descent, which may require many iterations depending on the condition number of $X^\top X$.

### Gradient and Hessian for multiple linear regression

For $L(\theta) = \frac{1}{n}\|y - X\theta\|^2$, we have $\nabla L = -\frac{2}{n}X^\top(y - X\theta)$ and $\nabla^2 L = \frac{2}{n}X^\top X$.

```python
# calculate the gradient and hessian for the linear function
# should differ case by case
def Gradient(X, y, theta):
    return -2 * X.T @ (y - X @ theta) / len(y)

def Hessian(X, y, theta):
    return 2 * X.T @ X / len(y)

def Error_Gradient(X, y, theta):
    return -X
```

```python
def Newton(X, y, epochs=1e3, tol=1e-8):
    epoch = 0
    if len(X.shape) == 1:
        print("yes")
        X = X.reshape(-1, 1)
        # add a column of ones into X
        intercept = np.ones(len(X)).reshape(-1,1)
        X, y = np.hstack((intercept, X)), np.array(y)
        # initialize a column of 1 as the intercept for the regression
        theta_t = np.zeros(np.shape(X)[1])
        for epoch in range(int(epochs)):
            print(epoch)
            dg = Gradient(X, y, theta_t)
            dH = Hessian(X, y, theta_t)

            theta_new = theta_t - np.linalg.inv(dH) @ dg
            delta = theta_new - theta_t
            theta_t = theta_new
            # if the norm of the difference is smaller than the threshold
            if sum(delta**2) < tol**2:
                return ({"Method": "Newton-Raphson", "nsteps": epoch+1, "b": theta_t[0], "omega": theta_t[1:]})
    else:
        print("no")
        # add a column of ones into X for the intercept
        intercept = np.ones(len(X)).reshape(-1,1)
        X, y = np.hstack((intercept, X)), np.array(y)
        theta_t = np.zeros(np.shape(X)[1])
        for epoch in range(int(epochs)):
            dg = Gradient(X, y, theta_t)
            dH = Hessian(X, y, theta_t)

            theta_new = theta_t - np.linalg.inv(dH) @ dg
            delta = theta_new - theta_t
            theta_t = theta_new
            if sum(delta**2) < tol**2:
                return ({"Method": "Newton-Raphson", "nsteps": epoch+1, "b": theta_t[0], "omega": theta_t[1:]})
    return ({"Method": "Newton-Raphson", "nsteps": epoch+1, "b": theta_t[0], "omega": theta_t[1:]})
```

## Part 3: Adaptive Optimizers -- SGD, Adagrad, RMSprop, Momentum, Adam

Below we define five optimizer classes that all expose a `Delta(dg)` method returning the parameter update $\Delta\theta$ given the current gradient $dg$:

| Optimizer | Key idea |
|---|---|
| **GD** | Fixed step: $\Delta = -\eta\, g$ |
| **Adagrad** | Per-coordinate adaptive rate using accumulated squared gradients |
| **RMSprop** | Exponential moving average of squared gradients (fixes Adagrad's decaying rate) |
| **Momentum** | Exponential moving average of gradients (smooths oscillations) |
| **Adam** | Combines momentum (first moment) and RMSprop (second moment) with bias correction |

```python

class GD():
    def __init__(self, eta=1e-3):
        self.eta = eta

    def Delta(self, dg):
        return -self.eta * dg


class Adagrad():
    def __init__(self, eta=1e-3, epsilon=1e-8):
        self.eta = eta
        self.epsilon = epsilon
        self.past_g = 0

    def Delta(self, dg):
        self.past_g += dg**2
        inv_G = np.diag((self.past_g**(1/2)+self.epsilon)**(-1))
        return -self.eta * inv_G @ dg

class RMSprop():
    def __init__(self, eta=1e-3, gamma = 0.9, epsilon=1e-8):
        self.eta = eta
        self.gamma = gamma
        self.epsilon = epsilon
        self.v = 0

    def Delta(self, dg):
        self.v = self.gamma * self.v + (1-self.gamma)*dg*dg
        inv_V = np.diag((self.v**(1/2)+self.epsilon)**(-1))
        return -self.eta * inv_V @ dg


class Momentum():
    def __init__(self, eta=1e-3, gamma = 0.9):
        self.eta = eta
        self.gamma = gamma
        self.m = 0

    def Delta(self, dg):
        self.m = self.gamma*self.m - (1-self.gamma)*dg
        return self.eta * self.m

class Adam():
    def __init__(self, eta=1e-3, beta1=0.9, beta2=0.999, epsilon=1e-8):
        self.m, self.v = 0,0
        self.beta1 = beta1
        self.beta2 = beta2
        self.epsilon = epsilon
        self.eta = eta
        self.t = 0

    def Delta(self, dg):
        self.t += 1
        self.m = self.beta1*self.m - (1-self.beta1)*dg
        self.v = self.beta2*self.v + (1-self.beta2)*dg**2
        hat_m = self.m/(1-self.beta1**self.t)
        hat_v = self.v/(1-self.beta2**self.t)
        inv_hat_V = np.diag((hat_v**(1/2) + self.epsilon)**(-1))
        return self.eta * inv_hat_V @ hat_m
```

```python
def Gradient(X, y, theta):
    return -X.T @ (y - X @ theta) / len(y)

def gradient_descent(X, y, optimizer, epochs=1e5):
    intercept = np.ones(len(X)).reshape(-1,1)
    if len(X.shape) == 1:
        X = X.reshape(-1, 1)
        X, y = np.hstack((intercept, X)), np.array(y)
        theta_t = np.zeros(np.shape(X)[1])
        for epoch in range(int(epochs)):
            # calculate gradient
            dg = Gradient(X, y, theta_t)
            theta_t += optimizer.Delta(dg)

        return({"Optimizer": optimizer.__class__.__name__, "omega": np.squeeze(theta_t[1:]), "b": theta_t[0]})
```

```python
data = pd.read_csv("./data/Advertising.csv", index_col=0)
X, y = data["TV"].values, data["sales"].values


print(gradient_descent(X, y, GD(eta=1e-5), epochs=1e6))
print(gradient_descent(X, y, Adagrad(eta=1), epochs=1e4))
print(gradient_descent(X, y, RMSprop(eta=1e-4), epochs=1e5))
print(gradient_descent(X, y, Momentum(eta=1e-4), epochs=2e5))
print(gradient_descent(X, y, Adam(eta=1), epochs=1e3))
```

## Part 4: Optimizer Trajectories on Beale's Function

**Beale's function** is a classic nonconvex benchmark:

$$
f(x, y) = (1.5 - x + xy)^2 + (2.25 - x + xy^2)^2 + (2.625 - x + xy^3)^2.
$$

It has a global minimum at $(x^*, y^*) = (3, 0.5)$ and features a long, curved valley that challenges gradient-based optimizers. We run all five optimizers from the starting point $(1, 1)$ and visualize their trajectories on the contour plot.

This comparison reveals important differences:
- **Vanilla GD** takes many small steps along the valley floor.
- **Adagrad** and **Adam** adapt their step sizes per coordinate, enabling faster navigation through the curved valley.
- **Momentum** overshoots initially but then settles quickly.
- **RMSprop** can be sensitive to the learning rate on this landscape.

```python
Beale = lambda x, y: (1.5-x+x*y)**2 + (2.25-x+x*y**2)**2 + (2.625-x+x*y**3)**2

def Beale_Gradient(theta):
    x, y = theta
    dx = -12.75 + 3*y + 4.5*y**2 + 5.25*y**3 + 2*x*(3-2*y-y**2-2*y**3+y**4+y**6)
    dy = 6*x*(0.5+1.5*y+2.625*y**2+x*(-1/3-1/3*y-y**2+2/3*y**3+y**5))
    return np.array([dx, dy])

def Beale_Gradient_Descent(theta_t, optimizer, max_epochs=1e5, tol=1e-5):
    paths = [theta_t]
    for epoch in range(int(max_epochs)):
        dg = Beale_Gradient(theta_t)
        theta_new = theta_t+optimizer.Delta(dg)
        paths.append(theta_new)
        if sum((theta_new-theta_t)**2) < tol**2:
            return (np.squeeze(paths))
        theta_t = theta_new
    return (np.squeeze(paths))

start = np.array([1,1])
start_ = start.reshape(-1,1)

GD_path = Beale_Gradient_Descent(start, GD(eta=1e-2))
Adagrad_path = Beale_Gradient_Descent(start, Adagrad(eta=1))
RMSprop_path = Beale_Gradient_Descent(start, RMSprop(eta=1e-3))
Momentum_path = Beale_Gradient_Descent(start, Momentum(eta=1e-1))
Adam_path = Beale_Gradient_Descent(start, Adam(eta=1))

print({'GD': len(GD_path)})
print({'Adagrad': len(Adagrad_path)})
print({'RMSprop': len(RMSprop_path)})
print({'Momentum': len(Momentum_path)})
print({'Adam': len(Adam_path)})

print({"nstep": len(RMSprop_path), "eta": 1e-2, "RMSprop": RMSprop_path[-1]})
```

## Part 5: SGD vs. Mini-Batch SGD

**Stochastic gradient descent (SGD)** updates parameters using the gradient from a single sample, while **mini-batch SGD** averages over a small batch. Using a mini-batch reduces the variance of gradient estimates, leading to smoother convergence trajectories at the cost of more computation per step.

Below we generate synthetic data from $y = 2x + 5 + \varepsilon$ and compare:
- **SGD** (batch size 1): noisy, zig-zagging path toward the optimum.
- **Mini-batch SGD** (batch size 32): smoother trajectory with fewer total updates.

```python
import time
from sklearn.utils import shuffle

np.random.seed(4012)
N = 100
x = np.linspace(-1, 1, N)
y = 2 * x + 5 + np.random.randn(N) / 100
x, y = shuffle(x, y, random_state=4012)

def gradient(x, y, w, b):
    """Gradient of squared loss for a single (or batch of) sample(s)."""
    return -x * (y - (w * x + b)), -(y - (w * x + b))
```

```python
# SGD algorithm
def SGD(inits, X, y, lr=1e-3, n_iter=10000, eta=1e-12):
    w, b = inits
    w_list, b_list = [w], [b]
    for i in range(n_iter):
        for j in range(len(y)):
            grad_w, grad_b = gradient(X[j], y[j], w, b)
            w -= lr*grad_w
            b -= lr*grad_b
            w_list.append(w)
            b_list.append(b)
            if (w-w_list[-2])**2 + (b-b_list[-2])**2 < eta:
                return w_list, b_list
    return w_list, b_list

start_time = time.time()
SGD_w, SGD_b = SGD([-4, 0], x, y, lr=0.02, eta=1e-8)
SGD_time = time.time() - start_time
print({"Method": "SGD with eta=1e-8", "niter": len(SGD_w), "time": SGD_time, "w": SGD_w[-1], "b":SGD_b[-1]})


start_time = time.time()
SGD_w, SGD_b = SGD([-4, 0], x, y, lr=0.02, eta=1e-12)
SGD_time = time.time() - start_time
print({"Method": "SGD with eta=1e-12", "niter": len(SGD_w), "time": SGD_time, "w": SGD_w[-1], "b":SGD_b[-1]})
```

```python
# Mini-batch SGD
def MbSGD(inits, X, y, lr=1e-3, n_iter=10000, bs=32, eta=1e-8):
    ind = np.arange(len(y))
    w, b = inits
    w_list, b_list = [w], [b]
    batch_indices = [ind[i:(i + bs)] for i in range(0, len(y), bs)]
    for i in range(n_iter):
        for indices in batch_indices:
            grad_sum_w, grad_sum_b = 0, 0
            for j in indices:
                grad_w, grad_b = gradient(X[j], y[j], w, b)
                grad_sum_w += grad_w
                grad_sum_b += grad_b
            w -= lr * grad_sum_w / len(indices)  # average gradient over batch
            b -= lr * grad_sum_b / len(indices)
            w_list.append(w)
            b_list.append(b)
            if (w - w_list[-2])**2 + (b - b_list[-2])**2 < eta:
                return w_list, b_list
    return w_list, b_list

start_time = time.time()
MbSGD_w, MbSGD_b = MbSGD([-4, 0], x, y, lr=0.02, eta=1e-8)
MbSGD_time = time.time() - start_time
print({"Method": "Mini-batch SGD with eta=1e-8", "niter": len(MbSGD_w), "time": MbSGD_time, "w": MbSGD_w[-1], "b": MbSGD_b[-1]})

start_time = time.time()
MbSGD_w, MbSGD_b = MbSGD([-4, 0], x, y, lr=0.02, eta=1e-12)
MbSGD_time = time.time() - start_time
print({"Method": "Mini-batch SGD with eta=1e-12", "niter": len(MbSGD_w), "time": MbSGD_time, "w": MbSGD_w[-1], "b": MbSGD_b[-1]})
```

### SGD vs. Mini-batch trajectory comparison

The contour plot below shows the MSE loss surface for the 2-parameter linear model. The red path (SGD) exhibits noisy oscillations since each update uses only one data point. The black path (mini-batch) is much smoother because it averages gradients over a batch of 32 samples.

```python
w = np.linspace(-6, 10, 20)
b = np.linspace(-2, 10, 20)
Wgrid, Bgrid = np.meshgrid(w, b)
MSE = 0
for xs, ys in zip(x, y):
    MSE += (ys-(Wgrid*xs+Bgrid))**2/N/2

plt.figure(figsize=(10, 6))
cp = plt.contour(Wgrid, Bgrid, MSE, levels=25, linestyles="dashed")
plt.xlabel('w')
plt.ylabel('b')
plt.title("MSE of SGD")
plt.plot(SGD_w, SGD_b, color='r', label='SGD')
plt.plot(MbSGD_w, MbSGD_b, color="black", label = "Mini-batch")
plt.legend(framealpha=1, frameon=True)
plt.savefig("./figures/SGD_vs_Minibatch.png", dpi=1000)
plt.show()
```

### Vectorized mini-batch SGD

The implementation below replaces the inner Python loop over batch samples with a vectorized NumPy computation, which is significantly faster for larger batch sizes.

```python
def MbSGD_Mat(inits, X, y, lr=1e-3, n_iter=10000, bs=32, eta=1e-8):
    ind = np.arange(len(y))
    w, b = inits
    w_list, b_list = [w], [b]
    batch_indices = [ind[i:(i+bs)] for i in range(0, len(y), bs)]
    for _ in range(n_iter):
        for indices in batch_indices:
            grad_w, grad_b = gradient(X[indices], y[indices], w, b)
            w -= lr*np.mean(grad_w)
            b -= lr*np.mean(grad_b)
            w_list.append(w)
            b_list.append(b)
            if (w-w_list[-2])**2 + (b-b_list[-2])**2 < eta:
                return w_list, b_list
    return w_list, b_list


start_time = time.time()
MbSGD_w, MbSGD_b = MbSGD([-4, 0], x, y, lr=0.02, eta=1e-8)
MbSGD_time = time.time() - start_time
print({"Method": "Mini-batch SGD with eta=1e-8", "niter": len(MbSGD_w), "time": MbSGD_time, "w": MbSGD_w[-1], "b":MbSGD_b[-1]})
```

## Summary

**Key takeaways from this notebook:**

- **Vanilla GD** is reliable on convex landscapes but can be slow when the loss surface is ill-conditioned (elongated contours).
- **Newton's method** converges in very few steps by leveraging curvature information, but requires computing and inverting the Hessian -- prohibitively expensive for large models.
- **Adaptive optimizers** (Adagrad, RMSprop, Adam) rescale the learning rate per coordinate, which is especially beneficial on loss surfaces with different curvatures along different directions. On Beale's function, Adam and Adagrad converge in roughly 200 steps versus 1500+ for vanilla GD.
- **Momentum** smooths out oscillations and accelerates movement along consistent gradient directions.
- **SGD vs. mini-batch SGD:** single-sample SGD produces noisy trajectories; mini-batching reduces variance at the cost of more computation per update. In practice, batch sizes of 32--256 strike a good balance.
